```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Menentukan URL target
url_audio = "https://chomsky.info/audionvideo/"
response_audio = requests.get(url_audio)
soup_audio = BeautifulSoup(response_audio.text, "html.parser")

# Mengambil semua elemen list (li)
titles = soup_audio.find_all("li")
data = []

for item in titles:
    teks = item.get_text(strip=True)
    if teks:
        data.append(teks)

# Menyimpan data ke format CSV menggunakan pandas
if data:
    df = pd.DataFrame(data, columns=["title"])
    df.to_csv("chomsky.csv", index=False)
    print(f"SUKSES! Berhasil mengambil {len(data)} data dan menyimpannya ke 'chomsky.csv'.")
else:
    print("Tidak ada data yang ditemukan.")
```
